## Part 1: 記憶體分析

```python
# 創建示例數據
np.random.seed(42)
n_rows = 100000

df = pd.DataFrame({
    'int_col': np.random.randint(0, 1000, n_rows),
    'float_col': np.random.random(n_rows),
    'str_col': np.random.choice(['A', 'B', 'C', 'D'], n_rows),
    'category_col': np.random.choice(['X', 'Y', 'Z'], n_rows),
    'bool_col': np.random.choice([True, False], n_rows)
})

print("=== 原始數據記憶體使用 ===")
print(f"\n{df.info()}")
print(f"\n總記憶體使用: {df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")
```
